```python
# Healthcare Bronze Layer - Raw Data Ingestion

base_path = "/Volumes/workspace/healthcare/raw_data"

# Bronze tables location
bronze_path = "workspace.healthcare"

# List of source datasets
tables = ["patients", "doctors", "appointments", "billing"]

for table_name in tables:

    print(f"Processing: {table_name}")

    # Read raw CSV data
    df = (
        spark.read
        .option("header", "true")
        .option("inferSchema", "true")
        .csv(f"{base_path}/{table_name}")
    )

    # Write data into Delta table
    (
        df.write
        .format("delta")
        .mode("overwrite")
        .saveAsTable(f"{bronze_path}.bronze_{table_name}")
    )

    print(f"Bronze table created: bronze_{table_name}")

print("Bronze ingestion completed!")
```

```python
# Verify Bronze tables

tables = [
    "bronze_patients",
    "bronze_doctors",
    "bronze_appointments",
    "bronze_billing"
]

for table_name in tables:

    print(f"\nChecking table: {table_name}")

    df = spark.table(f"workspace.healthcare.{table_name}")

    print("Total Records:", df.count())

    df.show(5, truncate=False)
```
