```python
from pyspark.sql import functions as F

# Storage location in Databricks Volume
base_path = "/Volumes/workspace/healthcare/raw_data"

# 1. Patients data
patients = [
    (101, "Ravi", 36, "Male", "Hyderabad"),
    (102, "Sita", 28, "Female", "Bangalore"),
    (103, "Kumar", 52, "Male", "Chennai"),
    (104, "Anjali", 41, "Female", "Hyderabad"),
    (105, "Rahul", 30, "Male", "Pune"),
    (106, "Priya", 26, "Female", "Mumbai")
]

patients_df = spark.createDataFrame(
    patients,
    ["PatientID", "PatientName", "Age", "Gender", "City"]
)

# 2. Doctors data
doctors = [
    (201, "Dr. Arun", "Cardiology", 12),
    (202, "Dr. Meena", "Neurology", 8),
    (203, "Dr. Ravi", "Orthopedics", 15),
    (204, "Dr. Priya", "Pediatrics", 6),
    (205, "Dr. Kiran", "Dermatology", 10)
]

doctors_df = spark.createDataFrame(
    doctors,
    ["DoctorID", "DoctorName", "Department", "Experience"]
)

# 3. Appointments data
appointments = [
    (1001, 101, 201, "2026-10-01", "Completed"),
    (1002, 102, 202, "2026-10-02", "Completed"),
    (1003, 103, 203, "2026-10-03", "Pending"),
    (1004, 104, 204, "2026-10-04", "Completed"),
    (1005, 105, 205, "2026-10-05", "Cancelled"),
    (1006, 106, 201, "2026-10-06", "Completed"),
    (1007, 101, 202, "2026-10-07", "Pending"),
    (1008, 103, 203, "2026-10-08", "Completed")
]

appointments_df = spark.createDataFrame(
    appointments,
    ["AppointmentID", "PatientID", "DoctorID", "AppointmentDate", "Status"]
)

# 4. Billing data
billing = [
    (5001, 1001, 2500.0, "Paid"),
    (5002, 1002, 1800.0, "Paid"),
    (5003, 1003, 3000.0, "Pending"),
    (5004, 1004, 1500.0, "Paid"),
    (5005, 1005, 1000.0, "Cancelled"),
    (5006, 1006, 2200.0, "Paid"),
    (5007, 1007, 1800.0, "Pending"),
    (5008, 1008, 3500.0, "Paid")
]

billing_df = spark.createDataFrame(
    billing,
    ["BillingID", "AppointmentID", "Amount", "PaymentStatus"]
)

# Save all four datasets as CSV files
datasets = {
    "patients": patients_df,
    "doctors": doctors_df,
    "appointments": appointments_df,
    "billing": billing_df
}

for name, df in datasets.items():
    (
        df.write
        .mode("overwrite")
        .option("header", "true")
        .csv(f"{base_path}/{name}")
    )

    print(f"{name}.csv data saved successfully")

print("Healthcare synthetic data creation completed!")
```
